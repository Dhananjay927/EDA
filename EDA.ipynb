2- Explain overfitting and underfitting in ML. How can you detect and prevent
them?
Hint: Discuss bias-variance tradeoff, cross-validation, and regularization techniques

- **Overfitting and Underfitting** are common problems in machine learning that describe how well a model generalizes to new, unseen data, based on how well it learned the training data. Both are related to the **bias-variance tradeoff**.

- Bias-Variance Tradeoff:

  *   **Bias:** Refers to the simplifying assumptions made by a model to make the target function easier to learn (it is also called the traning error). High bias models are typically too simple and may lead to **underfitting**.
  *   **Variance:** Refers to the model's sensitivity to small fluctuations in the training data. (it is also called the testing error) High variance models are typically too complex and may lead to **overfitting**.
  *   The tradeoff means that reducing bias often increases variance, and vice-versa. The goal is to find a balance.

- Overfitting:

  *   **Definition:** Occurs when a model learns the training data too well, including its noise and outliers. It performs excellently on the training data but poorly on unseen test data.
  *   **Characteristics:** The model is too complex for the given data. It essentially memorizes the training examples rather than learning the underlying patterns.
  *   **Analogy:** Imagine studying for a test by memorizing every single question and answer from the practice exams. You'd ace the practice exams, but if the real test has new questions (even on the same topics), you'd struggle.
  *   **Detection:**
      *   **Performance Gap:** A large difference between training accuracy (high) and test/validation accuracy (low).
      *   **Complex Model:** The model has too many parameters or features relative to the amount of training data.
  *   **Prevention:**
      *   **More Training Data:** Provide the model with a larger and more diverse dataset.
      *   **Feature Selection/Engineering:** Reduce the number of features or create more meaningful ones to simplify the model.
      *   **Regularization:** Add a penalty term to the loss function during training, discouraging the model from assigning too much importance to any single feature (e.g., L1, L2 regularization).
      *   **Cross-Validation:** Helps estimate the model's performance on unseen data by splitting the data into multiple folds and training/testing on different combinations. This gives a more robust evaluation of generalization error.
      *   **Early Stopping:** Stop training the model when its performance on a validation set starts to degrade, even if its performance on the training set is still improving.
      *   **Ensemble Methods:** Combine multiple models (e.g., Random Forests, Gradient Boosting) to reduce variance.
      *   **Dropout (for Neural Networks):** Randomly deactivate a percentage of neurons during training, preventing co-adaptation and forcing the network to learn more robust features.

- Underfitting:

  *   **Definition:** Occurs when a model is too simple to capture the underlying patterns in the training data. It performs poorly on both training and test data.
  *   **Characteristics:** The model has high bias. It fails to learn the relationships between features and the target variable.
  *   **Analogy:** Imagine trying to understand a complex novel by only reading the first sentence of every chapter. You'd miss most of the plot and characters, and wouldn't be able to answer questions about the book accurately.
  *   **Detection:**
      *   **Low Performance:** The model has low accuracy or high error rates on both the training and test sets.
      *   **Simple Model:** The model is too basic for the complexity of the problem.
  *   **Prevention:**
      *   **More Complex Model:** Use a more sophisticated model with more parameters (e.g., more layers in a neural network, more complex decision tree).
      *   **More Features:** Add more relevant features to the dataset, or perform more extensive feature engineering.
      *   **Reduce Regularization:** If regularization was applied, reduce its strength or remove it.
      *   **Increase Training Time:** Train the model for more epochs if it hasn't converged yet (though be careful not to overfit).
      *   **Feature Scaling/Preprocessing:** Ensure features are properly scaled or preprocessed, as some algorithms are sensitive to feature scales.
      *   **Remove Noise:** Clean the data by removing obvious errors or irrelevant noise that might be confusing a simple model.

3- How would you handle missing values in a dataset? Explain at least three
methods with examples.
Hint: Consider deletion, mean/median imputation, and predictive modeling.

- Missing values are like gaps in your data. You can't just ignore them because they can mess up your analysis. Here are three main ways to deal with them, each with its own pros and cons:

1. Deletion:

    *   **What it is:** Simply remove any rows or columns that contain missing values.
    *   **When to use:**
        *   **Row Deletion (Listwise Deletion):** If only a very small percentage of your data (rows) have missing values, and these missing values are randomly distributed. If you have many rows, losing a few won't hurt much.
        *   **Column Deletion:** If an entire column has too many missing values to be useful (e.g., 70-80% missing).
    *   **Example:** Imagine a survey where only 2 out of 100 people skipped one question. You might just delete those 2 responses.

2. Imputation:

    *   **What it is:** Fill in the missing values with a substitute. This is more sophisticated than deletion because it tries to preserve more data.
    *   **Common Imputation Techniques:**
        *   **Mean/Median Imputation:** Replace missing values in a column with the mean (average) or median (middle value) of that column. Median is better for skewed data or when outliers are present.
            *   **Example:** If you have ages of students and a few are missing, you might fill them in with the average age of the other students. If incomes are missing and some are very high, median income is a better choice.
        *   **Mode Imputation:** For categorical data, replace missing values with the most frequent category (the mode).
            *   **Example:** If `favorite_color` is missing, you'd fill it with the most common favorite color in your dataset.


  3. Predictive Modeling:

      *   **What it is:** Use a machine learning model to predict the missing values based on other features in the dataset.
      *   **When to use:** When missing values are not random and there's a strong relationship between the missing data and other variables. It's more accurate but also more complex.
      *   **Example:** If you're missing a person's 'salary', you could build a model using their 'education_level', 'years_experience', and 'job_title' to predict what their salary would likely be. You could use algorithms like K-Nearest Neighbors (KNN), Regression, or even more advanced methods.


4- What is an imbalanced dataset? Describe two techniques to handle it
(theoretical + practical).
Hint: Discuss SMOTE, Random Under/Oversampling, and class weights in models.


- Imbalanced Dataset:

  - An **imbalanced dataset** is one where the number of examples in one class (the majority) is much larger than in other classes (the minority). For example, 95% 'normal' transactions and 5% 'fraudulent' transactions. This is a problem because standard ML algorithms often focus on the majority class, leading to poor performance on the crucial minority class.

  - **Example Scenario:** A model predicting fraud might achieve 99% accuracy by always predicting 'normal' if only 1% of transactions are fraudulent. It effectively ignores the fraud cases.

#### Techniques to Handle Imbalanced Datasets

We'll demonstrate with a synthetic imbalanced dataset first:

**1. Synthetic Imbalanced Data Generation**

```python
import pandas as pd
import numpy as np
from collections import Counter
from sklearn.datasets import make_classification
import matplotlib.pyplot as plt
import seaborn as sns

# Generate a synthetic imbalanced dataset (95% majority, 5% minority)
X, y = make_classification(n_samples=1000, n_features=2, n_informative=2,
                           n_redundant=0, n_repeated=0, n_classes=2,
                           n_clusters_per_class=1, weights=[0.95, 0.05],
                           flip_y=0, random_state=42)

# Convert to DataFrame for visualization
df = pd.DataFrame(X, columns=['feature_1', 'feature_2'])
df['target'] = y

print("Original dataset class distribution:", Counter(y))

plt.figure(figsize=(8, 6))
sns.scatterplot(x='feature_1', y='feature_2', hue='target', data=df, palette='coolwarm', s=100, alpha=0.7)
plt.title('Imbalanced Dataset Visualization')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend(title='Class')
plt.show()
```

---

- Technique 1: Resampling (Random Undersampling and Oversampling)

  - **Theoretical Explanation:**

    *   **Undersampling:** Reduces the number of samples in the majority class to match the minority class. Simple, but can lose valuable information.
    *   **Oversampling:** Duplicates instances from the minority class to balance the dataset. Simple, but can lead to overfitting if not carefully used.

  - **Practical Python Example (using `imblearn`):**

```python

# =========================================================
# RANDOM OVERSAMPLING (MANUAL METHOD)
# =========================================================

# Separate majority and minority classes
majority = df[df['target'] == 0]
minority = df[df['target'] == 1]

# Oversample minority class
minority_oversampled = minority.sample(
    n=len(majority),
    replace=True,
    random_state=42
)

# Combine both classes
df_oversampled = pd.concat([majority, minority_oversampled])

# Shuffle dataset
df_oversampled = df_oversampled.sample(frac=1, random_state=42)

# Check distribution
print("\nAfter Random Oversampling:")
print(Counter(df_oversampled['target']))

# Visualize oversampled dataset
plt.figure(figsize=(8,6))
sns.scatterplot(
    x='feature_1',
    y='feature_2',
    hue='target',
    data=df_oversampled,
    palette='coolwarm',
    s=100,
    alpha=0.7
)
plt.title("Random Oversampling")
plt.show()


# =========================================================
# RANDOM UNDERSAMPLING (MANUAL METHOD)
# =========================================================

# Randomly reduce majority class
majority_undersampled = majority.sample(
    n=len(minority),
    random_state=42
)

# Combine classes
df_undersampled = pd.concat([majority_undersampled, minority])

# Shuffle dataset
df_undersampled = df_undersampled.sample(frac=1, random_state=42)

# Check distribution
print("\nAfter Random Undersampling:")
print(Counter(df_undersampled['target']))

# Visualize undersampled dataset
plt.figure(figsize=(8,6))
sns.scatterplot(
    x='feature_1',
    y='feature_2',
    hue='target',
    data=df_undersampled,
    palette='coolwarm',
    s=100,
    alpha=0.7
)
plt.title("Random Undersampling")
plt.show()
```

---

- Technique 2: SMOTE (Synthetic Minority Over-sampling Technique)

  - **Theoretical Explanation:** SMOTE creates *new, synthetic* minority samples instead of just duplicating existing ones. It does this by taking a minority sample, finding its neighbors, and creating new samples along the lines connecting them. This helps to create a more diverse and realistic minority class, reducing overfitting risks.

  - **Practical Python Example (using `imblearn`):**

```python
from imblearn.over_sampling import SMOTE

print("Original class distribution:", Counter(y))

# SMOTE Oversampling
smote = SMOTE(random_state=42)
X_smote, y_smote = smote.fit_resample(X, y)
print("\nSMOTE Oversampling:", Counter(y_smote))

df_smote = pd.DataFrame(X_smote, columns=['feature_1', 'feature_2']); df_smote['target'] = y_smote
plt.figure(figsize=(8, 6)); sns.scatterplot(x='feature_1', y='feature_2', hue='target', data=df_smote, palette='coolwarm', s=100, alpha=0.7)
plt.title('SMOTE Resampling Visualization'); plt.show()
```

These methods help ensure your models don't ignore the less frequent, but often more important, minority classes.

5- Why is feature scaling important in ML? Compare Min-Max scaling and Standardization. Hint: Explain impact on distance-based algorithms (e.g., KNN, SVM) and gradient descent.

- Why Feature Scaling is Crucial in Machine Learning (Interview Focus):-

Feature scaling is a preprocessing step that transforms the range of independent features or variables. It's vital for many ML algorithms to ensure they perform optimally and converge efficiently.

**Key Reasons for Importance:**

1.  **Impact on Distance-Based Algorithms (e.g., KNN, SVM, K-Means):**
    *   **Problem:** These algorithms calculate distances between data points. If features have vastly different scales (e.g., age vs. income), the feature with the largest range will disproportionately influence the distance, making other features seem less important. This can lead to biased results.
    *   **Solution:** Scaling ensures all features contribute equally to the distance calculation, preventing features with larger numerical values from dominating.
    *   **Example:** Without scaling, a small change in 'income' (e.g., $1000) might seem more significant than a large change in 'age' (e.g., 10 years), simply because 'income' values are much larger.

2.  **Impact on Gradient Descent Based Algorithms (e.g., Linear/Logistic Regression, Neural Networks, Gradient Boosting with custom objectives):**
    *   **Problem:** Gradient descent algorithms minimize a cost function by iteratively adjusting model parameters. If features are unscaled, the cost function's contours will be elongated ellipses. The optimization process will take small, zig-zagging steps, making convergence very slow.
    *   **Solution:** Scaling creates a more spherical cost function contour. This allows gradient descent to take more direct and larger steps towards the minimum, leading to faster convergence and more efficient training.

3.  **Regularization Techniques (e.g., L1/L2 Regularization):**
    *   **Problem:** Regularization penalizes large coefficients to prevent overfitting. If features are unscaled, the penalty might unfairly shrink the coefficients of features with smaller scales, while larger-scaled features get less penalty, leading to biased regularization.
    *   **Solution:** Scaling ensures that regularization applies equally to all features, preventing bias towards features with smaller numerical ranges.

**In essence: Scaling prevents features with larger numerical ranges from unduly influencing the model, leading to better performance, faster convergence, and fairer regularization.**

### Comparing Min-Max Scaling (Normalization) and Standardization

Both Min-Max Scaling (also known as Normalization) and Standardization (also known as Z-score normalization) are fundamental techniques, but they differ in their approach and suitable applications.

**Min-Max Scaling** rescales features to a fixed range, typically between 0 and 1. Its formula is `X_scaled = (X - X_min) / (X_max - X_min)`. This method is straightforward and preserves the original distribution shape of the data. It's particularly useful when algorithms require inputs within a specific bounded range, such as certain activation functions in neural networks. However, its main drawback is its high sensitivity to outliers; extreme values can compress the majority of the data into a very small range, potentially losing important distinctions.

**Standardization**, on the other hand, transforms data to have a mean of 0 and a standard deviation of 1. Its formula is `X_scaled = (X - μ) / σ` (where μ is the mean and σ is the standard deviation). Unlike Min-Max scaling, Standardization does not bound the features to a specific range, but it is much more robust to outliers. While it's often associated with data that follows an approximately Gaussian (normal) distribution, it generally performs well even with non-normal data. It's preferred for algorithms that assume normally distributed data (like LDA) or those sensitive to feature scales but not to fixed ranges (such as SVMs, PCA, and K-Means).

**In summary:** Choose **Min-Max Scaling** when you need features to be within a specific range and are confident that outliers are not a major issue or have been handled. Opt for **Standardization** when robustness to outliers is important, when your data is approximately normally distributed, or when the algorithm does not require bounded inputs but benefits from features having similar scales and distributions around zero.

6- Compare Label Encoding and One-Hot Encoding. When would you prefer one over the other?
Hint: Consider categorical variables with ordinal vs. nominal relationships.

### Label Encoding

**What it is:** Label Encoding converts categorical (non-numeric) data into numerical format. Each unique category in a column is assigned a unique integer value.

**Example:** If you have a 'Size' column with categories ['Small', 'Medium', 'Large'], Label Encoding might convert them to [0, 1, 2] respectively.

**Advantages:**
*   Simple and memory-efficient: It only adds one numerical column to the dataset.

**Disadvantages:**
*   **Implies Order (Ordinality):** The main problem is that it introduces an artificial ordinal relationship between categories. For example, `Large (2)` is numerically greater than `Medium (1)`, which is greater than `Small (0)`. While this is correct for 'Size', it would be problematic for 'Color' (e.g., `Red=0, Green=1, Blue=2` does not mean Blue > Green > Red).
*   Can mislead algorithms: Many machine learning algorithms (especially those that calculate distances or assume continuous numerical data, like linear models or SVMs) might misinterpret these numerical labels as having a meaningful order or magnitude, leading to incorrect assumptions and poorer model performance.

### One-Hot Encoding

**What it is:** One-Hot Encoding transforms categorical variables into a binary (0 or 1) numerical format. For each category in a column, it creates a *new* binary column. If a row belongs to that category, the value in the new column is 1; otherwise, it's 0.

**Example:** If you have a 'Color' column with categories `['Red', 'Green', 'Blue']`:
*   A new column 'Color_Red' is created.
*   A new column 'Color_Green' is created.
*   A new column 'Color_Blue' is created.
*   If an observation is 'Red', `Color_Red=1`, `Color_Green=0`, `Color_Blue=0`.

**Advantages:**
*   **No Implied Order:** It effectively removes the problem of ordinality, as each category is treated as independent.
*   Suitable for nominal data: Algorithms can correctly interpret the categories without false assumptions about their relationships.

**Disadvantages:**
*   **Increased Dimensionality:** For categorical variables with many unique categories, One-Hot Encoding can create a large number of new columns. This can lead to the "curse of dimensionality," increasing memory usage, computation time, and potentially causing issues for some models.
*   **Multicollinearity (Dummy Variable Trap):** When creating `N` new columns for `N` categories, one column is redundant (its value can be inferred from the others). This can cause multicollinearity in linear models. To avoid this, it's common practice to drop one of the created columns (e.g., if you have 'Red', 'Green', 'Blue', you only create 'Color_Red' and 'Color_Green'; if both are 0, it implies 'Blue').

### When to Prefer One Over the Other

The choice between Label Encoding and One-Hot Encoding primarily depends on the nature of your categorical variable:

1.  **Prefer Label Encoding when:**
    *   **The categorical variable has an intrinsic ordinal relationship:** This means the categories have a clear, meaningful order. Examples include 'Size' (Small < Medium < Large), 'Education Level' (High School < Bachelor's < Master's), 'Rating' (Bad < Good < Excellent).
    *   **The algorithm can handle ordinal data:** Some tree-based algorithms (like Decision Trees, Random Forests, Gradient Boosting Machines) can sometimes work well with Label Encoded ordinal features because they can split on thresholds that make sense for the ordering.

2.  **Prefer One-Hot Encoding when:**
    *   **The categorical variable has a nominal relationship:** This means the categories have no inherent order. Examples include 'Color' (Red, Green, Blue), 'City' (New York, London, Tokyo), 'Animal Type' (Dog, Cat, Bird).
    *   **The algorithm is sensitive to ordinality:** This includes most linear models (Linear Regression, Logistic Regression), Support Vector Machines (SVMs), K-Nearest Neighbors (KNN), and Neural Networks. These algorithms would incorrectly interpret the numerical magnitude of Label Encoded nominal features.
    *   **The number of unique categories is relatively small:** To avoid the curse of dimensionality, One-Hot Encoding is more practical when the categorical feature doesn't have too many distinct values.

**Rule of thumb:** When in doubt, especially for nominal categories, One-Hot Encoding is generally the safer choice to prevent your model from making false assumptions about category relationships.

8- Titanic Dataset Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Titanic dataset
titanic_df = pd.read_csv('/content/titanic.csv')

print("Titanic dataset loaded successfully. First 5 rows:")
display(titanic_df.head())

print("\nDataset Information:")
titanic_df.info()
```

#### a) Compare the survival rates based on passenger class (Pclass). Which class had the highest survival rate, and why do you think that happened?

```python
# Calculate survival rates by Pclass
pclass_survival = titanic_df.groupby('Pclass')['Survived'].mean().reset_index()
pclass_survival['Survived_Rate'] = pclass_survival['Survived'] * 100

print("Survival Rates by Passenger Class:")
display(pclass_survival.sort_values(by='Survived_Rate', ascending=False))

# Visualize survival rates by Pclass
plt.figure(figsize=(8, 6))
sns.barplot(x='Pclass', y='Survived_Rate', data=pclass_survival, palette='viridis')
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate (%)')
plt.ylim(0, 100)
plt.xticks(ticks=[0, 1, 2], labels=['1st Class', '2nd Class', '3rd Class'])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

#### Survival Rates by Passenger Class

**Findings:**
First Class passengers had a significantly higher survival rate (approximately 63%) compared to Second Class (around 47%) and Third Class (about 24%) passengers.

**Why this happened:**
This disparity is largely attributed to the "women and children first" protocol, combined with the social hierarchy on board.

1.  **Prioritization in Evacuation:** First-class cabins were typically located on higher decks, closer to the lifeboats. This, coupled with the prevailing social etiquette that prioritized first-class passengers during an emergency, meant they had preferential access to lifeboats.
2.  **Wealth and Influence:** First-class passengers were generally wealthier and more influential, which may have afforded them better treatment and quicker responses from the crew during the crisis.
3.  **Cabin Location:** As mentioned, their cabins were often on the upper decks, making it easier and quicker to reach the lifeboats compared to third-class passengers, who were often housed on lower decks and had more obstacles to navigate to reach the upper evacuation points.

In essence, social status and logistical advantages played a crucial role in determining survival outcomes.

#### Survival Rates by Age Group

**Findings:**
Yes, children (Age < 18) had a better chance of survival (approximately 54%) compared to adults (about 38%).

**Why this happened:**
This observation is primarily due to the humanitarian principle of "women and children first" that was reportedly followed during the Titanic evacuation.

1.  **Ethical Priority:** In disaster situations, there is often an unwritten, and sometimes explicit, rule to save the most vulnerable first. Children, being less capable of self-preservation in a crisis, were given priority for lifeboat spots.
2.  **Cultural Norms:** The societal values of the time heavily emphasized the protection of children, leading to crew members and other passengers actively assisting children into lifeboats.

#### a) How do flight prices vary with the days left until departure? Identify any exponential price surges and recommend the best booking window.

```python
# The 'days_left' column is already available in the dataset, so no need to recalculate.

# Group by days left and calculate average price
avg_price_by_days_left = flight_df.groupby('days_left')['price'].mean().reset_index()

# Plotting the variation of price with days left
plt.figure(figsize=(12, 7))
sns.lineplot(x='days_left', y='price', data=avg_price_by_days_left)
plt.title('Average Flight Price vs. Days Left Until Departure')
plt.xlabel('Days Left Until Departure')
plt.ylabel('Average Price')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xlim(0, avg_price_by_days_left['days_left'].max())
plt.xticks(np.arange(0, avg_price_by_days_left['days_left'].max() + 1, 15))
plt.show()
```

#### Flight Price Variation and Booking Window

**Findings:**
The plot shows a general trend: flight prices tend to be higher when booking very close to the departure date (e.g., within 7-10 days) and when booking too far in advance (e.g., beyond 60-90 days). There's often a 'sweet spot' where prices are lowest.

**Exponential Price Surges:**
We can typically observe an **exponential surge in prices as the departure date approaches very closely (e.g., the last 7-14 days)**. Airlines implement dynamic pricing, and as seats fill up and demand increases for remaining seats, algorithms rapidly increase prices. This is particularly true for business travelers or last-minute emergencies who have less price sensitivity.

**Recommended Best Booking Window:**
Based on general trends in flight pricing (which this plot likely reflects):

*   **Sweet Spot:** The optimal booking window usually falls between **30 to 90 days before departure**. During this period, airlines have adjusted their initial high prices if demand was low, but haven't yet initiated the last-minute price hikes.
*   **Avoid:**
    *   **Too Early (more than 90-120 days):** Prices can be higher initially as airlines test demand and haven't fully optimized their pricing. There's also a higher chance of schedule changes.
    *   **Too Late (less than 14-21 days):** Prices surge significantly due to high demand for remaining seats and catering to urgent travel.

**Key Factors Influencing:**

*   **Airline Algorithms:** Sophisticated models adjust prices based on demand, time until departure, competitor pricing, and historical data.
*   **Load Factor:** As a flight fills up, remaining seats become more expensive.
*   **Traveler Type:** Airlines aim to capture high-paying last-minute business travelers with inflated prices, while enticing leisure travelers with earlier discounts.
*   **Seasonality/Events:** Peak travel seasons, holidays, and major events will shift this window and generally raise prices.

```python
# Convert 'salary' to numerical for correlation calculation if it's categorical string
# First, check unique values for 'salary' and 'sales'
print("Unique 'salary' values:", hr_df['salary'].unique())
print("Unique 'sales' values:", hr_df['sales'].unique())

# 'left' column is our target variable for attrition (assuming 1 means left, 0 means stayed)
# Let's ensure it's numeric.

# Convert 'salary' to numerical using Label Encoding for correlation
salary_mapping = {'low': 0, 'medium': 1, 'high': 2}
hr_df['salary_encoded'] = hr_df['salary'].map(salary_mapping)

# The 'overtime' column does not exist in this dataset, so we will skip its encoding and visualization.
hr_df['overtime_encoded'] = 0 # Defaulting to 0 as the column is not present

# Calculate correlation matrix for numerical features and 'left'
# Exclude 'sales' as it's a categorical string, and other non-numeric IDs
numerical_cols = hr_df.select_dtypes(include=np.number).columns.tolist()

# Ensure 'left' is in numerical_cols if it exists. If not, error handling or assumption needs review.
# For this dataset, 'left' is already numeric (0/1).

# Select relevant columns for correlation, including the encoded ones
correlation_cols = ['satisfaction_level', 'last_evaluation', 'number_project', 'average_montly_hours',
                    'time_spend_company', 'Work_accident', 'promotion_last_5years', 'left']

if 'salary_encoded' in hr_df.columns:
    correlation_cols.append('salary_encoded')

# Filter for only existing columns to avoid errors if some are missing
correlation_cols = [col for col in correlation_cols if col in hr_df.columns]

correlation_matrix = hr_df[correlation_cols].corr()

print("\nCorrelation with 'left' (Attrition):")
display(correlation_matrix['left'].sort_values(ascending=False))

# Visualize top correlated features
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix[['left']].drop('left').sort_values(by='left', ascending=False), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation of Features with Employee Attrition (left)')
plt.show()

# Visualizations for key drivers
# Satisfaction Level vs Attrition
plt.figure(figsize=(8, 6))
sns.boxplot(x='left', y='satisfaction_level', data=hr_df)
plt.title('Attrition vs. Satisfaction Level')
plt.xlabel('Left Company (0=No, 1=Yes)')
plt.ylabel('Satisfaction Level')
plt.show()

# Salary vs Attrition
plt.figure(figsize=(8, 6))
sns.countplot(x='salary', hue='left', data=hr_df, order=['low', 'medium', 'high'], palette='coolwarm')
plt.title('Attrition vs. Salary')
plt.xlabel('Salary')
plt.ylabel('Number of Employees')
plt.show()
```
